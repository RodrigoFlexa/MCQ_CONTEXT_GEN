# Piloto de contexto e questões — v3
Entrada: tópico, subtópico e frase integral do especialista. Fluxo por seções e lotes, com retomada.
Execute no ambiente do servidor com as dependências de `requirements.txt`. As chamadas usam Azure e o corpus instalado.

```python
import os, sys, subprocess, json
from pathlib import Path
ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'pipeline/fase_3/topicos.py').is_file())
PYTHON = os.environ.get('DOCUMENT_PYTHON', sys.executable)
OUTPUT = 'pipeline/qa_research/saida_fluxo_v3/02_02_piloto_01'
COMMAND = [PYTHON, '-u', '-m', 'pipeline.qa_research.v3', '--item', '02_02', '--manifest', 'dataset/professional_corpora_v2/manifest.json', '--output', OUTPUT]
```

```python
# Mostra a configuração sem acessar a API.
subprocess.run(COMMAND + ['--plan'], cwd=ROOT, check=True)
```

```python
# Testa a configuração e os modelos com chamadas curtas.
subprocess.run(COMMAND + ['--check'], cwd=ROOT, check=True)
```

```python
# Executa ou retoma o piloto. O estado real aparece no terminal e em resumo.json.
result = subprocess.run(COMMAND, cwd=ROOT, check=False)
print('Código de saída:', result.returncode)
```
